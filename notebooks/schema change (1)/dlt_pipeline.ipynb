```python
source = spark.conf.get("source")
```

```python
import dlt

# Bronze Layer
@dlt.table(
name="bronze_user"
)
def bronze_table():
  return (
    spark.readStream.format("cloudFiles") \
      .option("cloudFiles.format", "parquet") \
      .option("cloudFiles.inferColumnTypes", True) \
      .load(f"{source}")
  )


 
```

```python
#Silver Layer
@dlt.table(
  name=f"silver_user"
)
@dlt.expect("schema_update_rule", "_rescued_data is null")
def silver_table():
  return(
    dlt.read_stream('bronze_user')
  )
```
